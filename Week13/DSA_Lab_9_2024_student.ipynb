You are tasked with implementing several core operations of a Binary Search Tree (BST) using a provided code template. The BST is constructed from nodes defined by the `BinaryTreeNode` class, where each node contains a key, an associated value, and pointers to its left and right children.

Your implementation will focus on fundamental BST operations including:
- **Checking emptiness:** Determining whether the tree contains any nodes.
- **Recursive insertion:** Inserting new key-value pairs into the BST using a recursive strategy.
- **Traversals:** Displaying the tree's contents using Preorder, Inorder, and Postorder traversals.

### Structured Test Case

Run the following code and see the result:
- If all tests pass, you'll see 'All test cases passed' (which is good!).
- If a test fails, you'll see an AssertionError indicating which test case failed and why.

```python
class TestBinarySearchTree:
    def test_is_empty(self):
        bst = BinarySearchTree()
        assert bst.is_empty() == True
        bst.insert(10, "Root")
        assert bst.is_empty() == False

    def test_insert_recursive(self):
        bst = BinarySearchTree()
        bst.insert(10, "Root")
        bst.insert(5, "Left")
        bst.insert(15, "Right")
        
        assert bst._root._key == 10
        assert bst._root._leftChild._key == 5
        assert bst._root._rightChild._key == 15

    def test_traverse(self, capsys):
        bst = BinarySearchTree()
        bst.insert(10, "Root")
        bst.insert(5, "Left")
        bst.insert(15, "Right")
        bst.traverse()
        captured = capsys.readouterr()
        assert "Preorder:" in captured.out
        assert "Inorder:" in captured.out
        assert "Postorder:" in captured.out

    def test_tree_search(self):
        bst = BinarySearchTree()
        bst.insert(10, "Root")
        bst.insert(5, "Left")
        bst.insert(15, "Right")
        
        assert bst.treeSearch(10) == "Root"
        assert bst.treeSearch(5) == "Left"
        assert bst.treeSearch(15) == "Right"
        assert bst.treeSearch(20) == None

    def test_find_min(self):
        bst = BinarySearchTree()
        bst.insert(10, "Root")
        bst.insert(5, "Left")
        bst.insert(15, "Right")
        assert bst.findMin_re(bst._root) == (5, "Left")

    def test_find_max(self):
        bst = BinarySearchTree()
        bst.insert(10, "Root")
        bst.insert(5, "Left")
        bst.insert(15, "Right")
        assert bst.findMax_re(bst._root) == (15, "Right")
```

```python
def test_bst():
    passed_cases = 0
    failed_cases = 0
    test = TestBinarySearchTree()
    if test.test_is_empty():
        print("Test case test_is_empty passed")
        passed_cases += 1
    else:
        print("Test case test_is_empty failed")
        failed_cases += 1

    if test.test_insert_recursive():
        print("Test case test_insert_recursive passed")
        passed_cases += 1
    else:
        print("Test case test_insert_recursive failed")
        failed_cases += 1
    
    if test.test_traverse():
        print("Test case test_traverse passed")
        passed_cases += 1
    else:
        print("Test case test_traverse failed")
        failed_cases += 1

    if test.test_tree_search():
        print("Test case test_tree_search passed")
        passed_cases += 1
    else:
        print("Test case test_tree_search failed")
        failed_cases += 1

    if test.test_find_min():
        print("Test case test_find_min passed")
        passed_cases += 1
    else:
        print("Test case test_find_min failed")
        failed_cases += 1

    if test.test_find_max():
        print("Test case test_find_max passed")
        passed_cases += 1
    else:
        print("Test case test_find_max failed")
        failed_cases += 1

    print(f"\nSummary:")
    print(f"Passed cases: {passed_cases}")
    print(f"Failed cases: {failed_cases}")
    
# Run the tests
if __name__ == "__main__":
    test_bst()
```
